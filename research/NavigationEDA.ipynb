```python
import csv
from datetime import datetime

# --- File paths ---
input_file = r"F:\zizo\RTKCorrection\data\navigation\BaseStationNav.24P"  # RINEX navigation file
output_file = r"F:\zizo\RTKCorrection\data\navigation\GPS_RefernceNavigation_parameters.csv"

# --- Processing ---
data_rows = []

with open(input_file, "r") as f:
    lines = f.readlines()

# Skip first 5 header lines, real data starts at line 6 (index 5)
for line in lines[5:]:
    # Each satellite entry starts with PRN info (first 2 chars are system, next 2 are PRN number)
    prn_str = line[0:3].strip()  # Example: "G04"
    if not prn_str:  # Skip empty lines
        continue

    try:
        gnss_type = prn_str[0]
        prn_number = int(prn_str[1:])  # Remove 'G' and convert to int
    except ValueError:
        continue  # Skip if not valid

    # Extract date/time from positions according to RINEX nav format
    year = int(line[4:8].strip())
    month = int(line[9:11].strip())
    day = int(line[12:14].strip())
    hour = int(line[15:17].strip())
    minute = int(line[18:20].strip())
    second = float(line[21:23].strip())

    # Convert to datetime object (UTC)
    epoch_datetime = datetime(year, month, day, hour, minute, int(second))
    epoch_str = epoch_datetime.strftime("%Y-%m-%d %H:%M:%S")

    # Extract Clock parameters (positions follow standard RINEX nav file format)
    clock_bias = float(line[23:42].strip())
    clock_drift = float(line[42:61].strip())
    clock_drift_rate = float(line[61:80].strip())

    if gnss_type == 'G':
        data_rows.append([gnss_type, prn_number, epoch_str, clock_bias, clock_drift, clock_drift_rate])

# --- Save to CSV ---
with open(output_file, "w", newline="") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["PRN","Number", "Epoch (UTC)", "Clock Bias (s)", "Clock Drift (s/s)", "Clock Drift Rate (s/s²)"])
    writer.writerows(data_rows)

print(f"✅ Extraction complete. Data saved to {output_file}")
```
